```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import dense_rank, col
from pyspark.sql.window import Window
import os
```

```python
problem_name = os.path.basename(os.path.abspath("."))
```

```python
spark = SparkSession.builder.appName(problem_name).getOrCreate()
```

```python
salary_df = salaries = spark.read\
.option("header", "true")\
.option("infer_schema", "true")\
.format("csv")\
.load("files/salary_data.txt")
```

```python
salary_df.withColumn("rank_by_salary", dense_rank().over(Window.orderBy(col("salary").desc())))\
.filter(col("rank_by_salary")==3)\
.show()
```

```python
spark.stop()
```
