```python
"""
This notebook is used for setting basic utils 
for other notebooks where we solve leetcode SQL problems.
"""
import requests
from bs4 import BeautifulSoup
```

```python
def description_from_url(url):
    """
    This function returns description of leetcode problem when its url is passed.
    """
    problem_name = url.partition("description")[0].strip("/").split("/")[-1]
    data = {
    "operationName":"questionData",
    "variables":{"titleSlug":f"{problem_name}"},
    "query":"""query questionData($titleSlug: String!) {\n  question(titleSlug: $titleSlug) {\n    questionId\n    questionFrontendId\n    boundTopicId\n    title\n    titleSlug\n    content\n    translatedTitle\n    translatedContent\n    isPaidOnly\n    difficulty\n    likes\n    dislikes\n    isLiked\n    similarQuestions\n    contributors {\n      username\n      profileUrl\n      avatarUrl\n      __typename\n    }\n    langToValidPlayground\n    topicTags {\n      name\n      slug\n      translatedName\n      __typename\n    }\n    companyTagStats\n    codeSnippets {\n      lang\n      langSlug\n      code\n      __typename\n    }\n    stats\n    hints\n    solution {\n      id\n      canSeeDetail\n      __typename\n    }\n    status\n    sampleTestCase\n    metaData\n    judgerAvailable\n    judgeType\n    mysqlSchemas\n    enableRunCode\n    enableTestMode\n    envInfo\n    libraryUrl\n    __typename\n  }\n}\n"""
           }
    post_url = "https://leetcode.com/graphql/"
    timeout_seconds = 300
    json_response = requests.post(post_url, json=data, timeout=timeout_seconds).json()
    html_content = json_response["data"]["question"]["content"]
    return BeautifulSoup(html_content, "html.parser").get_text()
```

```python
def write_string_to_csvfile(string_data, file):
    """
    This function is used to write string data params to file, provided CSV file is passed.
    """
    filename = file.partition(".csv")[0]
    with open(f"{filename}_converted.csv", "w+", encoding="utf-8") as f:
        f.write(string_data)
        print(f"Completed writing data to {filename}_converted.csv")
```

```python
def convert_ascii_to_csv(file):
    """
    This function is used to convert ASCII text from a file to CSV data, 
    and CSV data is written to a file with '_converted' suffixed.
    """
    with open(file, encoding="utf-8") as f:
        data = f.readlines()
    new_data = []
    for line in data:
        if line.startswith("+"):
            continue
        line = line.strip().strip("|")
        new_data.append(",".join([col.strip() for col in line.split("|")]))
    file_data = "\n".join(new_data).strip()
    write_string_to_csvfile(file_data, file)
```
