```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg
import os
```

```python
problem_name = os.path.basename(os.path.abspath("."))
```

```python
spark = SparkSession.builder.appName(problem_name).getOrCreate()
```

```python
employees = spark.read\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .format("csv")\
    .load("files/employees.txt")
```

```python
employees.show()
```

```python
result_df = employees.groupBy(col("department"))\
    .agg(avg(col("salary")).alias("average_salary"))
```

```python
result_df.show(truncate=False)
```

```python
spark.stop()
```
