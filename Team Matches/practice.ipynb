```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
data = [["a"], ["b"], ["c"]]
```

```python
df = spark.createDataFrame(data, "team string")
```

```python
df.show()
```

```python
df.createOrReplaceTempView("teams")
```

```python
res_df = df.alias("t1").join(df.alias("t2"))\
    .filter("t1.team<t2.team")
```

```python
res_df.show()
```

```python
spark.sql(
    """
    select t1.team as T1, t2.team as T2 from teams t1 join teams t2
    where t1.team<t2.team
    """
).show()
```
