```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
data = [["a"], ["b"], ["c"]]
```

```python
df = spark.createDataFrame(data, "team string")
```

```python
df.show()
```

```python
res_df = df.alias("t1").join(df.alias("t2"))\
    .filter("t1.team<t2.team")
```

```python
res_df.show()
```
