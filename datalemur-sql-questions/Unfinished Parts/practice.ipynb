```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
parts_assembly = spark.read\
.option("delimiter", "\t")\
.option("header", "true")\
.option("inferSchema", "false")\
.schema("part string, finish_date string, assembly_step int")\
.format("csv")\
.load("parts_assembly.txt")
```

```python
parts_assembly.createOrReplaceTempView("parts_assembly")
```

```python
spark.sql(
    """
    select part, assembly_step from parts_assembly where finish_date is null
    """
).show(truncate=False)
```

```python
parts_assembly.filter("finish_date is null")\
.select("part", "assembly_step")\
.show(truncate=False)
```
